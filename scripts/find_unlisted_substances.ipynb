```python
import pandas as pd
import json
import re

```

```python
# Path to he substances JSON file
# Extracted substances from 'data/input/substances.json'
substances_file_path = 'substances.json'

with open(substances_file_path, 'r') as file:
    data = json.load(file)

print(data)
```

```python
all_values = [item for sublist in data.values() for item in sublist]
print(all_values)
```

```python
# Path to bnf UK Biobank lookups csv file
file_path = 'data/lkps/bnf_lkp.csv'

bnf_substances_df = pd.read_csv(file_path)
```

```python
# We only consider substances from BNF groups starting with '02' (Cardiovascular System)
# and '04' (Central Nervous System), as these are the focus of our analysis.
filtered_df = bnf_substances_df[
    bnf_substances_df['BNF_Presentation_Code'].str.startswith(('02', '04'), na=False)
]
```

```python
# Filter records that do not contain any value from all_values as a whole word in BNF_Chemical_Substance
excluded_records_df =filtered_df[
    ~filtered_df['BNF_Chemical_Substance'].apply(
        lambda cell: any(re.search(rf'\b{value}\b', str(cell), flags=re.IGNORECASE) for value in all_values)
    )
]
```

```python
excluded_substances = sorted(excluded_records_df['BNF_Chemical_Substance'].unique())
```

```python
len(excluded_substances)
```

```python
with open('excluded_substances.json', 'w') as jsonfile:
    json.dump(sorted(set(excluded_substances)), jsonfile, indent=4)
```
